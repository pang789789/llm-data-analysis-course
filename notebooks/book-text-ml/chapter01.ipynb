### 실습 3. 분석에 필요한 컬럼만 선택하기

원본 데이터의 모든 컬럼이 이후 분석에 필요한 것은 아닙니다.

이번 과정에서는 다음 컬럼을 먼저 선택합니다.

```python
# 1. 분석에 필요한 8개 핵심 컬럼 리스트 정의
selected_columns = [
    '순위', '판매상품 ID', '상품명', '판매가', 
    '인물', '출판사', '발행(출시)일자', '분야'
]

# 2. 원본 df에서 지정한 컬럼만 선택해 새 DataFrame(df_books) 생성
df_books = df[selected_columns].copy()

# 3. 선택된 컬럼 목록 확인
print("선택된 컬럼 목록:\n", df_books.columns)

# 4. 선택된 데이터의 상위 5개 행 확인
df_books.head()
```

## 3. 분석에 필요한 컬럼 선택

* **선택 컬럼 (8개):** `순위`, `판매상품 ID`, `상품명`, `판매가`, `인물`, `출판사`, `발행(출시)일자`, `분야`
* **목적:** 분석 목적에 적합한 핵심 데이터만 남겨 작업의 효율성 및 데이터 명확성 확보
* **결과 (`df_books.head()`):** 필요한 8개 컬럼만 추출된 신규 DataFrame 생성 완료

### 실습 4. 컬럼 이름을 사용하기 쉽게 정리하기

원본 컬럼 이름은 사람이 읽기에는 좋지만 코드에서 반복해서 사용하기에는 조금 길 수 있습니다.

이번 실습에서는 다음처럼 간단하게 정리할 수 있습니다.

```python
# 1. 변경할 컬럼명을 사전(dictionary) 형태 {기존이름: 새이름}으로 정의
rename_columns = {
    '판매상품 ID': '판매상품ID',
    '인물': '저자',
    '발행(출시)일자': '발행일'
}

# 2. rename() 함수를 이용해 컬럼명 변경 (inplace=True로 원본 df_books에 적용)
df_books.rename(columns=rename_columns, inplace=True)

# 3. 변경 후 컬럼 목록 확인
print("변경 완료된 컬럼 목록:\n", df_books.columns)
```

## 4. 컬럼명 정제

* **변경 내역:**
  - `판매상품 ID` $\rightarrow$ `판매상품ID` (공백 제거)
  - `인물` $\rightarrow$ `저자` (직관적인 명칭으로 변경)
  - `발행(출시)일자` $\rightarrow$ `발행일` (간결하게 정리)
* **목적:** 코드 작성 시 간결하고 직관적인 컬럼명을 사용하여 접근 편의성 확보
* **결과 (`df_books.columns`):** 8개 컬럼명 정제 완료

### 실습 5. 결측치 확인하기

결측치는 값이 비어 있는 데이터를 의미합니다.

데이터 분석에서는 결측치를 무조건 삭제하지 않습니다.

먼저 어디에 얼마나 있는지 확인하고, 분석 목적에 맞게 처리 방법을 결정합니다.

```python
# 1. 컬럼별 결측치(비어있는 값) 개수 확인
print("1. 처리 전 결측치 개수:\n", df_books.isna().sum())

# 2. 결측치 대체: 저자 -> '미상', 분야 -> '미분류'
df_books['저자'] = df_books['저자'].fillna('미상')
df_books['분야'] = df_books['분야'].fillna('미분류')

# 3. 결측치 대체 후 결과 확인
print("\n2. 처리 후 결측치 개수:\n", df_books.isna().sum())
```

## 5. 결측치 확인 및 대체

* **원칙:** 결측치를 무조건 삭제(`dropna()`)하지 않고 분석 목적에 맞게 대체
* **대체 기준:**
  - `저자` 결측치 $\rightarrow$ `'미상'`
  - `분야` 결측치 $\rightarrow$ `'미분류'`
* **결과 (`df_books.isna().sum()`):** 지정한 컬럼의 결측치 대체 완료 및 비어있는 값 0건 확인

### 실습 6. 판매가를 숫자로 바꾸기

Excel에서 판매가가 다음과 같이 저장되어 있을 수 있습니다.

```python
# 1. 변환 전 판매가 컬럼의 자료형(dtype) 확인
print("1. 변환 전 자료형:", df_books['판매가'].dtype)

# 2. 쉼표(',') 제거 후 정수형(int)으로 변환
df_books['판매가'] = df_books['판매가'].astype(str).str.replace(',', '').astype(int)

# 3. 변환 후 자료형 및 상위 5개 값 확인
print("\n2. 변환 후 자료형:", df_books['판매가'].dtype)
print("\n3. 변환 후 상위 5개 가격:\n", df_books['판매가'].head())

# 4. 숫자 데이터 요약 통계 확인 (평균, 최댓값, 최솟값 등)
df_books['판매가'].describe()
```

판매가 컬럼의 쉼표제거, 형변환!

```python
# 1. 쉼표(,) 제거 후 정수형(int)으로 변환
df_books['판매가'] = df_books['판매가'].astype(str).str.replace(',', '').astype(int)

# 2. 변환 후 자료형(dtype) 확인 (int64로 나오면 성공!)
print("변환 후 자료형:", df_books['판매가'].dtype)

# 3. 변환된 상위 5개 값 및 수치 통계 확인
print("\n상위 5개 값:\n", df_books['판매가'].head())
df_books['판매가'].describe()
```

## 6. 판매가 자료형 변환

* **문제점:** 쉼표(`,`)가 포함된 가격 정보가 문자열(`object`)로 읽혀 수치 계산 불가
* **전처리:** `str.replace(',', '')`로 쉼표 제거 후 `astype(int)`로 정수형 변환
* **결과 (`df_books['판매가'].dtype`):** `int64` 변환 완료 및 `describe()`를 통한 요약 통계값 검증 완료

### 실습 7. 발행일을 날짜형으로 바꾸기

원본 데이터의 발행일은 다음처럼 보일 수 있습니다.

```python
# 1. 변환 전 발행일 컬럼의 자료형 확인
print("1. 변환 전 자료형:", df_books['발행일'].dtype)

# 2. YYYYMMDD 형식을 날짜 자료형(datetime)으로 변환
df_books['발행일'] = pd.to_datetime(df_books['발행일'].astype(str), format='%Y%m%d')

# 3. 변환 후 자료형 및 상위 5개 값 확인
print("\n2. 변환 후 자료형:", df_books['발행일'].dtype)
print("\n3. 변환 후 상위 5개 날짜:\n", df_books['발행일'].head())
```

## 7. 발행일 날짜형 변환

* **전처리:** `pd.to_datetime()` 및 `%Y%m%d` 포맷 지정을 통해 문자열/숫자 데이터를 날짜형으로 변환
* **목적:** 연도, 월별 추이 분석 및 기간 필터링이 가능하도록 데이터 형식 표준화
* **결과 (`df_books['발행일'].dtype`):** `datetime64[ns]` 타입으로 변환 및 `YYYY-MM-DD` 형식 출력 확인 완료

### 실습 8. 문자열 앞뒤 공백 정리하기

상품명, 저자, 출판사, 분야는 문자열 데이터입니다.

문자열 앞뒤에 불필요한 공백이 있으면 같은 값도 서로 다른 데이터처럼 처리될 수 있습니다.

예를 들어 다음 두 값은 사람이 보기에는 같지만 문자열로는 다릅니다.

```python
# 1. 공백을 정리할 주요 문자열 컬럼 리스트 정의
string_columns = ['상품명', '저자', '출판사', '분야']

# 2. str.strip()을 사용해 각 컬럼의 앞뒤 불필요한 공백 제거
for col in string_columns:
    df_books[col] = df_books[col].astype(str).str.strip()

# 3. 공백 제거 후 상위 5개 값 확인
df_books[string_columns].head()
```

## 8. 문자열 공백 정제

* **대상 컬럼:** `상품명`, `저자`, `출판사`, `분야`
* **전처리:** `str.strip()` 함수를 활용하여 텍스트 앞뒤에 존재하는 불필요한 공백 일괄 제거
* **목적:** 동일한 단어/분야가 공백 차이로 인해 서로 다른 데이터로 인식되는 오류 방지

### 실습 9. 중복 데이터 확인하기

# Chapter 01. 베스트셀러 데이터 기본 전처리 요약

### 2. 주요 전처리 진행 내역
* **결측치 처리:** `저자` $\rightarrow$ '미상', `분야` $\rightarrow$ '미분류'로 대체 완료
