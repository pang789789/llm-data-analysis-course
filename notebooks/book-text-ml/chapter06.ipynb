# 실습 4. Streamlit의 재실행 방식 이해하기 및 데이터 로딩 함수 작성

> **학습 목표**
> - Streamlit 앱의 재실행(Rerun) 메커니즘을 이해하고, 불필요한 반복 데이터 로딩을 방지하는 `@st.cache_data` 캐시 기능을 습득합니다.
> - `book_bestseller_clean.csv` 파일의 필수 컬럼 유무를 검증하고 결측치를 정제하는 캐싱 기반 데이터 로드 함수를 작성합니다.

Streamlit 앱은 사용자가 버튼을 누르거나 입력값을 바꿀 때마다 파이썬 스크립트 전체가 처음부터 다시 실행됩니다.  
매번 CSV 파일을 새로 읽고 전처리하면 앱이 느려지므로, `@st.cache_data` 데코레이터를 사용하여 처리 결과를 메모리에 캐싱(저장)해 둡니다.

````python
import pandas as pd
import streamlit as st

DATA_PATH = "book_bestseller_clean.csv"

# @st.cache_data: 데이터 로딩 및 전처리 결과를 캐싱하여 앱 재실행 시 연산 속도를 최적화합니다.
@st.cache_data
def load_data():
    # 1. UTF-8 인코딩으로 CSV 데이터 불러오기
    df = pd.read_csv(DATA_PATH, encoding="utf-8-sig")
    
    # 2. 필수 컬럼('상품명', '분야') 존재 여부 검증
    required_columns = ["상품명", "분야"]
    missing_columns = [col for col in required_columns if col not in df.columns]
    if missing_columns:
        raise ValueError(f"필수 컬럼이 누락되었습니다: {missing_columns}")
    
    # 3. '상품명' 및 '분야' 컬럼 전처리 (결측치 처리 및 문자열 공백 제거)
    df["상품명"] = df["상품명"].fillna("").astype(str).str.strip()
    df["분야"] = df["분야"].fillna("미분류").astype(str).str.strip()
    
    # 4. 상품명이 비어있는 행 제거 후 인덱스 재정렬
    df = df[df["상품명"] != ""].reset_index(drop=True)
    return df

# 데이터 로드 함수 실행 및 결과 검증
df = load_data()
print("전처리 완료 데이터 크기 (행, 열):", df.shape)
print("\n[분야별 도서 수 상위 5개]")
print(df["분야"].value_counts().head(5))
````

**실습 결과 및 설명**
- `@st.cache_data` 데코레이터 덕분에 최초 1회만 파일 읽기/전처리가 실행되고, 이후 사용자 조작 시에는 캐시된 데이터를 즉시 재사용합니다.
- 필수 컬럼(`상품명`, `분야`) 검증 및 결측치(`미분류`) 처리를 포함하여 다음 단계인 머신러닝 학습 및 추천 로직에서 에러가 발생하지 않도록 안전하게 정제되었습니다.

# 실습 2. CSV 데이터 불러오기

> **학습 목표**
> - `app.py`와 동일한 폴더에 위치한 `book_bestseller_clean.csv` 데이터를 읽어오는 함수를 정의합니다.
> - 필수 컬럼(`상품명`, `분야`) 존재 여부를 검증하고 결측치를 처리하는 캐싱 함수를 작성합니다.

`app.py`와 전처리된 CSV 파일이 아래와 같이 같은 폴더에 존재해야 합니다.

```text
book-text-ml/
├─ app.py
└─ book_bestseller_clean.csv

````python
import pandas as pd
import streamlit as st

# 1. 전처리된 도서 데이터 파일 경로 지정
DATA_PATH = "book_bestseller_clean.csv"

# 2. @st.cache_data: 데이터 로딩 결과를 메모리에 캐싱하여 불필요한 재로딩 방지
@st.cache_data
def load_data():
    # 한글 깨짐 방지를 위해 utf-8-sig 인코딩으로 CSV 로드
    df = pd.read_csv(DATA_PATH, encoding="utf-8-sig")
    
    # 필수 컬럼('상품명', '분야') 존재 여부 검증
    required_columns = ["상품명", "분야"]
    missing_columns = [
        col for col in required_columns
        if col not in df.columns
    ]
    if missing_columns:
        raise ValueError(
            f"필수 컬럼이 없습니다: {missing_columns}"
        )
    
    # '상품명' 결측치 처리 및 문자열 공백 제거
    df["상품명"] = (
        df["상품명"]
        .fillna("")
        .astype(str)
        .str.strip()
    )
    
    # '분야' 결측치를 '미분류'로 채우고 문자열 공백 제거
    df["분야"] = (
        df["분야"]
        .fillna("미분류")
        .astype(str)
        .str.strip()
    )
    
    # 상품명이 빈 값인 행 제외 후 인덱스 재정렬
    df = df[df["상품명"] != ""].reset_index(drop=True)
    return df

# 3. 데이터 로딩 함수 정상 작동 테스트
df = load_data()
print("로딩된 총 도서 수:", len(df))
print(df[["상품명", "분야"]].head())
````

**실습 결과 및 설명**
- 앱 UI를 생성하기 전, 캐싱 기반 데이터 로딩 함수가 정상 작동하여 데이터를 잘 읽어오는지 먼저 확인합니다.
- 필수 컬럼 유무 검사 및 결측치 정제가 완료된 `df`를 반환합니다.

# 실습 3. Streamlit 캐시(Cache) 이해하기

> **학습 목표**
> - Streamlit의 스크립트 전체 재실행(Rerun) 특성을 이해합니다.
> - `@st.cache_data`와 `@st.cache_resource` 데코레이터의 역할 차이를 구분합니다.

Streamlit은 사용자가 버튼을 누르거나 입력값을 변경할 때마다 스크립트 전체를 다시 실행합니다.  
매번 CSV 파일에서 데이터를 다시 읽고 머신러닝 모델을 새로 학습하면 불필요한 연산이 반복되어 앱이 매우 느려집니다.

**Streamlit 캐시 2가지 핵심 데코레이터**
- `st.cache_data` : 데이터 로딩, 데이터프레임 변환 결과 등 **데이터** 재사용
- `st.cache_resource` : Vectorizer, 머신러닝 모델 객체, DB 연결 등 **자원(Resource)** 재사용

> *"자주 다시 만들 필요가 없는 결과를 메모리에 기억해 두고 재사용한다."*

````python
import streamlit as st

# 1. st.cache_data 예시: CSV 로딩 결과나 데이터프레임 등 '데이터'를 메모리에 캐싱
@st.cache_data
def get_sample_data():
    # 반복해서 파일이나 DB를 읽지 않고 캐시된 데이터를 재사용합니다.
    return {"status": "성공", "items": [1, 2, 3]}

# 2. st.cache_resource 예시: 머신러닝 모델, TF-IDF Vectorizer 등 '객체/자원'을 캐싱
@st.cache_resource
def get_sample_model():
    # 반복해서 모델을 재학습하지 않고 학습된 모델 객체를 메모리에 보관합니다.
    return "학습 완료된 머신러닝 모델 객체"

# 캐싱 데코레이터 적용 함수 동작 확인
data = get_sample_data()
model = get_sample_model()

print("데이터 캐싱 확인:", data)
print("모델 캐싱 확인:", model)
````

**실습 결과 및 설명**
- 데이터프레임이나 리스트 같은 결과물은 `@st.cache_data`를 붙여 보관합니다.
- TF-IDF Vectorizer 및 Naive Bayes 모델처럼 한 번 생성된 후 계속 사용해야 하는 머신러닝 객체는 `@st.cache_resource`를 붙여 관리합니다.

# 실습 4. 분류 모델 준비하기

> **학습 목표**
> - 도서 제목 기반으로 분야를 예측하기 위해 `TfidfVectorizer`와 `MultinomialNB`(나이브 베이즈) 모델을 학습하는 함수를 정의합니다.
> - `@st.cache_resource` 데코레이터를 사용하여 불필요한 반복 학습 없이 완성된 모델 객체를 메모리에 재사용합니다.

Chapter 04에서는 모델 평가를 위해 train/test 데이터셋을 나누었지만, 이번 과정은 평가가 끝난 후 시연용 앱을 만드는 단계이므로 사용 가능한 라벨 데이터를 활용해 최종 분류 모델을 학습합니다.

````python
import streamlit as st
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

# @st.cache_resource: Vectorizer, 모델 등 머신러닝 객체를 메모리에 캐싱하여 반복 재학습 방지
@st.cache_resource
def train_classifier():
    # 1. 전처리된 도서 데이터 로드
    df = load_data()
    
    # 2. '미분류' 항목을 제외한 학습용 라벨 데이터 추출
    train_df = df[df["분야"] != "미분류"].copy()
    
    # 3. 분류를 위해 서로 다른 분야가 최소 2개 이상 존재하는지 검증
    if train_df["분야"].nunique() < 2:
        raise ValueError("서로 다른 분야가 2개 이상 필요합니다.")
    
    # 4. 도서 제목 텍스트를 TF-IDF 행렬로 변환 (fit_transform)
    vectorizer = TfidfVectorizer()
    X = vectorizer.fit_transform(train_df["상품명"])
    y = train_df["분야"]
    
    # 5. Naive Bayes 분류 알고리즘 객체 생성 및 모델 학습 (fit)
    model = MultinomialNB()
    model.fit(X, y)
    
    return vectorizer, model

# 분류 모델 학습 및 객체 생성 테스트
classifier_vectorizer, classifier_model = train_classifier()
print("분류 모델 학습 완료!")
print("학습된 분야 목록:", classifier_model.classes_)
````

**실습 결과 및 설명**
- 도서 제목을 수치화하는 `TfidfVectorizer`와 분야를 분류하는 `MultinomialNB` 모델이 정상적으로 준비되었습니다.
- `@st.cache_resource` 데코레이터 덕분에 앱이 다시 실행되어도 매번 모델을 다시 학습하지 않고 기존에 만들어둔 모델 객체를 메모리에서 즉시 불러와 사용하게 됩니다.

사용자가 직접 입력한 제목을 이전에 학습시킨 분류 모델과 연결하여 예상 분야를 화면에 보여주는 로직을 작성합니다.

````python
import streamlit as st

# 1. 학습된 Vectorizer 및 분류 모델 객체 불러오기
classifier_vectorizer, classifier_model = train_classifier()

# 2. UI 구성: 섹션 헤더 및 도서 제목 입력창
st.header("1. 도서 분야 예측")
user_title = st.text_input(
    "도서 제목을 입력하세요",
    placeholder="예: 처음 배우는 파이썬 데이터 분석",
)

# 3. '분야 예측' 버튼 클릭 이벤트 처리
if st.button("분야 예측"):
    clean_title = user_title.strip()
    
    # 빈 입력값 예외 처리
    if not clean_title:
        st.warning("도서 제목을 입력해 주세요.")
    else:
        # 중요: 새 제목은 fit_transform()이 아닌 기존 단어 공간을 사용하는 transform() 적용
        title_vector = classifier_vectorizer.transform([clean_title])
        predicted_category = classifier_model.predict(title_vector)[0]
        st.success(f"예상 분야: {predicted_category}")
````

# 실습 7. 추천용 TF-IDF 준비하기

> **학습 목표**
> - 도서 추천을 위한 전체 도서 제목의 TF-IDF 피처 행렬을 생성하는 `build_recommender()` 함수를 정의합니다.
> - `@st.cache_resource`를 활용하여 추천용 Vectorizer와 TF-IDF 행렬 객체를 메모리에 보관 및 재사용합니다.

Chapter 05에서 작성한 코사인 유사도 기반 추천 기능을 앱에 연결합니다.  
분류 기능과 추천 기능은 모두 TF-IDF를 사용하지만 그 목적이 서로 다릅니다.

* **분류용 TF-IDF:** 입력된 제목 텍스트 기반으로 **도서 분야(Category)를 예측**
* **추천용 TF-IDF:** 전체 도서 제목 벡터 간 **텍스트 유사도를 상호 비교**

````python
import streamlit as st
from sklearn.feature_extraction.text import TfidfVectorizer

# @st.cache_resource: Vectorizer 및 TF-IDF 행렬 객체를 캐싱하여 반복 재계산 방지
@st.cache_resource
def build_recommender():
    # 1. 전처리 완료된 도서 데이터 로드
    df = load_data()
    
    # 2. 추천용 TF-IDF Vectorizer 객체 생성 및 전체 도서 제목 변환 (fit_transform)
    vectorizer = TfidfVectorizer()
    title_matrix = vectorizer.fit_transform(df["상품명"])
    
    return vectorizer, title_matrix

# 추천용 TF-IDF 구축 및 행렬 크기 확인
recommender_vectorizer, title_matrix = build_recommender()
print("추천용 TF-IDF 행렬 구축 완료!")
print("TF-IDF 행렬 크기 (도서 수, 단어 수):", title_matrix.shape)
````

**실습 결과 및 설명**
- 전체 도서 제목이 수치화된 `title_matrix` 행렬로 정상 변환되었습니다.
- `@st.cache_resource` 데코레이터를 적용하여 사용자가 선택지를 바꿀 때마다 매번 TF-IDF 행렬을 다시 계산하지 않고 메모리에서 즉시 꺼내어 사용합니다.

유사 도서 추천을 위해 전체 도서 목록 중에서 기준이 될 도서 한 권을 선택받는 UI를 작성합니다.

````python
import streamlit as st

# 1. 캐시된 도서 데이터 로드
df = load_data()

# 2. selectbox 표시용 포맷 함수 정의
# (화면에는 사람이 읽기 쉽게 표시하고, 내부적으로는 index 값을 사용)
def format_book(index):
    title = df.loc[index, "상품명"]
    if "저자" in df.columns:
        author = str(df.loc[index, "저자"])
        return f"{title} | {author}"
    return title

# 3. 도서 선택 selectbox UI 생성
selected_index = st.selectbox(
    "기준 도서를 선택하세요",
    options=df.index.tolist(),
    format_func=format_book,
)

# 선택된 도서 확인
print("선택된 도서 인덱스:", selected_index)
print("선택된 도서 제목:", df.loc[selected_index, "상품명"])
````

# 실습 9. Top 5 추천 연결하기

> **학습 목표**
> - 코사인 유사도 추천 로직을 `recommend_books()` 함수로 구현합니다.
> - 선택한 도서에 대해 자기 자신을 제외하고 유사도가 가장 높은 상위 5권의 도서를 추출합니다.
> - `st.components.v1.html`을 활용해 커스텀 고양이 디자인 UI로 추천 목록과 유사도를 출력합니다.

선택한 도서의 TF-IDF 벡터와 전체 도서의 TF-IDF 행렬 간 코사인 유사도를 계산하고, 상위 5권을 커스텀 고양이 카드 형태로 화면에 보여줍니다.

````python
import pandas as pd
import streamlit as st
import streamlit.components.v1 as components
from sklearn.metrics.pairwise import cosine_similarity

# 1. 코사인 유사도 기반 도서 추천 함수 정의
def recommend_books(df, title_matrix, selected_index, top_n=5):
    # 선택한 도서의 벡터 추출 및 전체 도서와의 유사도 계산
    selected_vector = title_matrix[selected_index]
    similarities = cosine_similarity(selected_vector, title_matrix).ravel()
    
    # 자기 자신은 추천 목록에서 제외 (-1 처리)
    similarities[selected_index] = -1
    
    # 유사도가 높은 상위 top_n개 인덱스 추출
    top_indices = similarities.argsort()[::-1][:top_n]
    
    # 결과 데이터프레임 생성
    display_columns = [
        col for col in ["상품명", "저자", "출판사", "분야"]
        if col in df.columns
    ]
    result = df.iloc[top_indices][display_columns].copy()
    result["유사도"] = similarities[top_indices].round(3)
    
    return result

# 2. 추천용 TF-IDF 행렬 불러오기
_, title_matrix = build_recommender()

# 3. UI 구성: 섹션 헤더 및 추천 실행 버튼
st.header("2. 비슷한 도서 추천")

if st.button("비슷한 도서 5권 추천"):
    # 추천 데이터 획득
    rec_df = recommend_books(df, title_matrix, selected_index, top_n=5)
    
    # 고양이 스타일 HTML 목록 생성
    items_html = ""
    for idx, row in rec_df.iterrows():
        title = row["상품명"]
        author = row.get("저자", "")
        author_str = f" | {author}" if author else ""
        score = f"{row['유사도']:.2f}"
        
        items_html += f"""
        <li class="cat-item">
          <div class="cat-item-left">
            <span class="cat-icon">🐱</span>
            <span>{title}{author_str}</span>
          </div>
          <div class="cat-item-right">
            <span>{score}</span>
            <span class="cat-icon">😸</span>
          </div>
        </li>
        """
    
    cat_ui_html = f"""
    <!DOCTYPE html>
    <html>
    <head>
      <style>
        body {{
          font-family: 'Pretendard', sans-serif;
          margin: 0;
          padding: 10px;
          background-color: transparent;
        }}
        .cat-list {{
          list-style: none;
          padding: 0;
          margin: 0;
          display: flex;
          flex-direction: column;
          gap: 10px;
        }}
        .cat-item {{
          display: flex;
          justify-content: space-between;
          align-items: center;
          background-color: #FFEFEF;
          border: 1.5px solid #FFD0D0;
          border-radius: 25px;
          padding: 12px 20px;
        }}
        .cat-item-left {{
          display: flex;
          align-items: center;
          gap: 10px;
          font-size: 14px;
          font-weight: 600;
          color: #4A3A31;
        }}
        .cat-item-right {{
          display: flex;
          align-items: center;
          gap: 6px;
          font-size: 14px;
          font-weight: bold;
          color: #6E5343;
          background: #FFE0BA;
          padding: 4px 12px;
          border-radius: 12px;
        }}
        .cat-icon {{
          font-size: 16px;
        }}
      </style>
    </head>
    <body>
      <ul class="cat-list">
        {items_html}
      </ul>
    </body>
    </html>
    """
    
    # Streamlit 화면에 고양이 커스텀 UI 렌더링
    components.html(cat_ui_html, height=320, scrolling=True)
````

**실습 결과 및 설명**
- 코사인 유사도로 계산된 추천 도서 Top 5 목록이 몽글몽글한 고양이 캡슐 모양 디자인으로 출력됩니다.
- 추천 결과(상품명, 저자, 유사도 점수)가 HTML/CSS로 깔끔하게 렌더링되어 사용성이 향상되었습니다.

# 실습 10. 최종 app.py 통합하기 및 앱 실행

> **학습 목표**
> - 지금까지 작성한 데이터 로딩, 캐싱, 분류 모델, 추천 모델 및 커스텀 UI 기능을 하나의 `app.py` 스크립트로 완성합니다.
> - 완성된 Streamlit 웹 애플리케이션을 전체적으로 테스트하고 동작을 확인합니다.

이제 전체 코드가 들어간 `app.py` 파일을 저장하고 터미널에서 실행해 봅니다.

````python
import pandas as pd
import sklearn
import streamlit as st
import streamlit.components.v1 as components
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.naive_bayes import MultinomialNB

# ---------------------------------------------------------
# 1. 앱 기본 설정 및 제목
# ---------------------------------------------------------
st.set_page_config(page_title="교보문고 베스트셀러 분석 앱", page_icon="📚")
st.title("📚 교보문고 베스트셀러 텍스트 분석 앱")
st.write("도서 제목 기반 **분야 예측**과 **유사 도서 추천** 기능을 제공합니다.")

DATA_PATH = "book_bestseller_clean.csv"

# ---------------------------------------------------------
# 2. 데이터 로딩 (캐싱)
# ---------------------------------------------------------
@st.cache_data
def load_data():
    df = pd.read_csv(DATA_PATH, encoding="utf-8-sig")
    required_columns = ["상품명", "분야"]
    missing_columns = [
        col for col in required_columns if col not in df.columns
    ]
    if missing_columns:
        raise ValueError(f"필수 컬럼이 없습니다: {missing_columns}")

    df["상품명"] = df["상품명"].fillna("").astype(str).str.strip()
    df["분야"] = df["분야"].fillna("미분류").astype(str).str.strip()
    df = df[df["상품명"] != ""].reset_index(drop=True)
    return df


try:
    df = load_data()
except Exception as e:
    st.error(f"데이터를 불러오는 중 오류가 발생했습니다: {e}")
    st.stop()

# ---------------------------------------------------------
# 3. 분류 모델 학습 (캐싱)
# ---------------------------------------------------------
@st.cache_resource
def train_classifier():
    train_df = df[df["분야"] != "미분류"].copy()
    if train_df["분야"].nunique() < 2:
        raise ValueError("서로 다른 분야가 2개 이상 필요합니다.")

    vectorizer = TfidfVectorizer()
    X = vectorizer.fit_transform(train_df["상품명"])
    y = train_df["분야"]

    model = MultinomialNB()
    model.fit(X, y)
    return vectorizer, model


try:
    classifier_vectorizer, classifier_model = train_classifier()
except Exception as e:
    st.warning(f"분류 모델 준비 중 경고: {e}")
    classifier_vectorizer, classifier_model = None, None

# ---------------------------------------------------------
# 4. 추천용 TF-IDF 구축 (캐싱)
# ---------------------------------------------------------
@st.cache_resource
def build_recommender():
    vectorizer = TfidfVectorizer()
    title_matrix = vectorizer.fit_transform(df["상품명"])
    return vectorizer, title_matrix


recommender_vectorizer, title_matrix = build_recommender()

# ---------------------------------------------------------
# 5. [기능 1] 도서 분야 예측 UI
# ---------------------------------------------------------
st.markdown("---")
st.header("1. 도서 분야 예측")

user_title = st.text_input(
    "도서 제목을 입력하세요",
    placeholder="예: 처음 배우는 파이썬 데이터 분석",
)

if st.button("분야 예측"):
    clean_title = user_title.strip()
    if not clean_title:
        st.warning("도서 제목을 입력해 주세요.")
    elif classifier_model is None:
        st.error("분류 모델이 준비되지 않았습니다.")
    else:
        title_vector = classifier_vectorizer.transform([clean_title])
        predicted_category = classifier_model.predict(title_vector)[0]
        st.success(f"예상 분야: **{predicted_category}**")

# ---------------------------------------------------------
# 6. [기능 2] 비슷한 도서 추천 UI (고양이 커스텀 디자인)
# ---------------------------------------------------------
st.markdown("---")
st.header("2. 비슷한 도서 추천")


def format_book(index):
    title = df.loc[index, "상품명"]
    if "저자" in df.columns:
        author = str(df.loc[index, "저자"])
        return f"{title} | {author}"
    return title


selected_index = st.selectbox(
    "기준 도서를 선택하세요",
    options=df.index.tolist(),
    format_func=format_book,
)


def recommend_books(df, title_matrix, selected_index, top_n=5):
    selected_vector = title_matrix[selected_index]
    similarities = cosine_similarity(selected_vector, title_matrix).ravel()
    similarities[selected_index] = -1

    top_indices = similarities.argsort()[::-1][:top_n]

    display_columns = [
        col for col in ["상품명", "저자", "출판사", "분야"] if col in df.columns
    ]
    result = df.iloc[top_indices][display_columns].copy()
    result["유사도"] = similarities[top_indices].round(3)
    return result


if st.button("비슷한 도서 5권 추천"):
    rec_df = recommend_books(df, title_matrix, selected_index, top_n=5)

    # 고양이 스타일 HTML/CSS UI 렌더링
    items_html = ""
    for idx, row in rec_df.iterrows():
        title = row["상품명"]
        author = row.get("저자", "")
        author_str = f" | {author}" if author else ""
        score = f"{row['유사도']:.2f}"

        items_html += f"""
        <li class="cat-item">
          <div class="cat-item-left">
            <span class="cat-icon">🐱</span>
            <span>{title}{author_str}</span>
          </div>
          <div class="cat-item-right">
            <span>{score}</span>
            <span class="cat-icon">😸</span>
          </div>
        </li>
        """

    cat_ui_html = f"""
    <!DOCTYPE html>
    <html>
    <head>
      <style>
        body {{
          font-family: 'Pretendard', sans-serif;
          margin: 0;
          padding: 10px;
          background-color: transparent;
        }}
        .cat-list {{
          list-style: none;
          padding: 0;
          margin: 0;
          display: flex;
          flex-direction: column;
          gap: 10px;
        }}
        .cat-item {{
          display: flex;
          justify-content: space-between;
          align-items: center;
          background-color: #FFEFEF;
          border: 1.5px solid #FFD0D0;
          border-radius: 25px;
          padding: 12px 20px;
        }}
        .cat-item-left {{
          display: flex;
          align-items: center;
          gap: 10px;
          font-size: 14px;
          font-weight: 600;
          color: #4A3A31;
        }}
        .cat-item-right {{
          display: flex;
          align-items: center;
          gap: 6px;
          font-size: 14px;
          font-weight: bold;
          color: #6E5343;
          background: #FFE0BA;
          padding: 4px 12px;
          border-radius: 12px;
        }}
        .cat-icon {{
          font-size: 16px;
        }}
      </style>
    </head>
    <body>
      <ul class="cat-list">
        {items_html}
      </ul>
    </body>
    </html>
    """

    components.html(cat_ui_html, height=340, scrolling=True)
````
