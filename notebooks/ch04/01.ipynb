```python
from pathlib import Path

PROJECT_ROOT = Path.cwd()
#print(PROJECT_ROOT)

DATA_DIR = PROJECT_ROOT / "data" / "raw"

print(DATA_DIR)
```

```python
# 현재 폴더를 확인한다.
current = Path.cwd().resolve()
project_root = current
print(current)

# 현재 폴더 안에 data 폴더가 있는지 확인한다.
is_data = current / "data"
print("현재 폴더에 data라는 하위 폴더가 존재 하는가? :",is_data.is_dir())

# 만약 data 폴더가 있다면
if is_data.is_dir():
#   현재 폴더를 프로젝트 루트로 결정한다.
    project_root = current
#   탐색을 종료한다.
    break

# 만약 data 폴더가 없다면
if not is_data.is_dir():
#   부모 폴더로 한 단계 올라간다.
    current = current.parent
# 다시 data 폴더가 있는지 확인한다.
# 이 과정을 반복한다.
#ㄴ> while True: 무한루트 사용
while True:
    if (current / "data").is_dir():
        project_root = current
        break

    current = current.parent
# 총 코드 5줄 만으로 완성

# 더 이상 위로 올라갈 수 없는데도
# 데이터 폴더를 찾지 못했다면
# 오류를 발생시킨다.
print("프로젝트 루트 :", project_root)
```

```python
from pathlib import Path

def get_project_root():
    current = Path.cwd().resolve()
    project_root = current

    while True:
        if (current / "data").is_dir():
            project_root = current
            break

        current = current.parent
        
    print("프로젝트 루트 :", project_root)
#   return "프로젝트 무한루트 경로"
    return project_root
```

```python
# 규칙을 모르는 사람을 위한 함수 추가
# get_data_dir 이라는 함수 만들기.
def get_data_dir():
#   return 프로젝트 루트 / "data" ;프로젝트 루트는 우ㅣ에 만들어둔 함수를 호출한다
    return get_project_root() / "data"
```
